### 4. Calculate the Sum of Digits
**Task**: Given a number, find the sum of its digits.  
**Demo Input**: `123`  
**Demo Output**: `6`  
(Explanation: `1+2+3=6`)

```python
demo_input = 123
demo_output = 0
for i in str(demo_input):
    demo_output += int(i)

demo_output
```

### 5. Find Factorial of a Number
**Task**: Calculate the factorial of a given number.  
**Demo Input**: `4`  
**Demo Output**: `24`  
(Explanation: `4! = 4*3*2*1 = 24`)

```python
demo_input = 4
demo_output = 1

for i in range(1,demo_input+1):
    demo_output *= i

demo_output
```

### 6. Count Occurrences of a Digit
**Task**: Count how many times a specific digit appears max in a number.  
**Demo Input**:  1233321 <br>
**Demo Output**: `3`  <br>
(Explanation: `3` appears 3 times in `1233321`)

```python
demo_input = 1233321
max_app = 0
checked = []
demo_output = str(demo_input)[0]

for i in str(demo_input):
    if i in checked:
        continue

    count = 0
    for j in str(demo_input):
        if i==j:
            count += 1

    if count>max_app:
        max_app = count
        demo_output = i

    checked.append(i)

int(demo_output)
```

### 7. Find the GCD of Two Numbers
**Task**: Find the greatest common divisor (GCD) of two numbers.  
**Demo Input**: 8 and 12 <br>
**Demo Output**: `4`

```python
input_1 = 8
input_2 = 12

lower = input_1 if (input_1<input_2) else input_2

demo_output = lower

for i in range(lower,0,-1):
    if (input_1%i==0) and (input_2%i==0):
        demo_output = i
        break

demo_output
```

### 8. Reverse a String
**Task**: Reverse the given string.  
**Demo Input**: `"aiquest"`  
**Demo Output**: `"tseuqia"`

```python
demo_input = "aiquest"
demo_output = ""

for i in range(len(demo_input)-1,-1,-1):
    demo_output += demo_input[i]

demo_output
```

### 9. Check Armstrong Number
**Task**: Check if a number is an Armstrong number (the sum of its digits raised to the power of the number of digits equals the number).  
**Demo Input**: `153`  
**Demo Output**: `153 is an Armstrong Number`  
(Explanation: \(1^3 + 5^3 + 3^3 = 153\))

```python
demo_input = 153
digits = len(str(demo_input))

gen_num = 0

for i in str(demo_input):
    gen_num += int(i)**digits

if gen_num == demo_input:
    print(f"{demo_input} is an Armstrong Number")
else:
    print(f"{demo_input} is not an Armstrong Number")
```

### 10. Generate a Pattern
**Task**: Print a pyramid pattern with `n` rows.  
**Demo input:** 4 <br>
**Output:** You will see the pyramid pattern with 4 rows

```python
n = 4
dif = 0

for i in range(1,n+1):
    print(" "*(n-i), "*"*(i+dif), sep="")
    dif += 1
```
